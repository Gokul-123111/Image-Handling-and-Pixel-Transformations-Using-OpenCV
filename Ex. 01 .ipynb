14. Read the image ('boy.jpg').

```python
img = cv2.imread('boy.jpg')
plt.imshow(img[:,:,::-1])
plt.show()
```

15. Adjust the brightness of the image.

```python
# Create a matrix of ones (with data type float64)
matrix_ones = np.ones(img.shape, dtype="uint8") * 50
```

16. Create brighter and darker images.

```python
img_brighter = cv2.add(img, matrix_ones)
img_darker = cv2.subtract(img, matrix_ones)
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python
plt.figure(figsize=(12,4))

plt.subplot(131)
plt.imshow(img[:,:,::-1])
plt.title("Original")

plt.subplot(132)
plt.imshow(img_darker[:,:,::-1])
plt.title("Darker")

plt.subplot(133)
plt.imshow(img_brighter[:,:,::-1])
plt.title("Brighter")

plt.show()
```

18. Modify the image contrast.

```python
img_higher1 = img * 1.1
img_higher2 = img * 1.2
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
img1 = cv2.convertScaleAbs(img, alpha=0.8)
img2 = cv2.convertScaleAbs(img, alpha=1.2)

plt.subplot(131)
plt.imshow(img[:,:,::-1])
plt.subplot(132)
plt.imshow(img1[:,:,::-1])
plt.subplot(133)
plt.imshow(img2[:,:,::-1])
plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
b, g, r = cv2.split(img)

plt.subplot(131)
plt.imshow(b, cmap='gray')
plt.subplot(132)
plt.imshow(g, cmap='gray')
plt.subplot(133)
plt.imshow(r, cmap='gray')
plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python

merged = cv2.merge((b, g, r))

plt.subplot(121)
plt.imshow(img[:,:,::-1])
plt.subplot(122)
plt.imshow(merged[:,:,::-1])
plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(hsv)

plt.subplot(131)
plt.imshow(h, cmap='gray')
plt.subplot(132)
plt.imshow(s, cmap='gray')
plt.subplot(133)
plt.imshow(v, cmap='gray')
plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
merged = cv2.merge((h, s, v))

plt.subplot(121)
plt.imshow(img[:,:,::-1])

plt.subplot(122)
plt.imshow(merged)
plt.show()
```
